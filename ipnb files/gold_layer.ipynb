# Access Data

```python
spark.conf.set("fs.azure.account.auth.type.nyctaxistorgeacc.dfs.core.windows.net", "OAuth")
spark.conf.set("fs.azure.account.oauth.provider.type.nyctaxistorgeacc.dfs.core.windows.net", "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider")
spark.conf.set("fs.azure.account.oauth2.client.id.nyctaxistorgeacc.dfs.core.windows.net", "app id")
spark.conf.set("fs.azure.account.oauth2.client.secret.nyctaxistorgeacc.dfs.core.windows.net", "secret value")
spark.conf.set("fs.azure.account.oauth2.client.endpoint.nyctaxistorgeacc.dfs.core.windows.net", "https://login.microsoftonline.com/directory id/oauth2/token")
```

**creating database**

```python
%sql
create database if not exists gold
```

# Data reading

**import libraries**

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df_trip = spark.read.format('parquet')\
                .option('inferSchema',True)\
                .option('header',True)\
                .load('abfss://silver@nyctaxistorgeacc.dfs.core.windows.net/trip_data_enriched/')
```

```python
display(df_trip)
```

```python

df_trip.write.format('delta') \
        .mode('append') \
        .saveAsTable('gold.trip_data_table')
```
